# Training loop

```python
import torch
import torch.nn as nn
import torch.optim as optim

# veri
x = torch.unsqueeze(torch.linspace(-1, 1, 100), dim=1)
y = 2 * x + 1

# model
model = nn.Linear(1, 1)

# loss
loss_fn = nn.MSELoss()

# optimizer
optimizer = optim.Adam(model.parameters(), lr=0.1)
```

### Tahmin et → hatayı ölç → öğren → kendini düzelt → tekrar et

```python
for epoch in range(100):

    # 1. forward (tahmin)
    y_pred = model(x)

    # 2. loss (hata)
    loss = loss_fn(y_pred, y)

    # 3. backward (gradient hesapla)
    optimizer.zero_grad()
    loss.backward()

    # 4. update (güncelle)
    optimizer.step()

    # 5. yazdır
    if epoch % 10 == 0:
        print(loss.item())
```
